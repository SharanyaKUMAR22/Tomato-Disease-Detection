```python
import os
import random
import shutil

dataset_path = "/content/TomatoDatasetV2/extracted/augmented_dataset"
eval_path = "/content/TomatoDatasetV2/evaluation_subset"

random.seed(42)

# 10% from every class
EVAL_FRACTION = 0.10

os.makedirs(eval_path, exist_ok=True)

class_counts = {}

for class_name in sorted(os.listdir(dataset_path)):

    class_path = os.path.join(dataset_path, class_name)

    if not os.path.isdir(class_path):
        continue

    images = [
        f for f in os.listdir(class_path)
        if f.lower().endswith(
            (".jpg", ".jpeg", ".png", ".bmp", ".webp")
        )
    ]

    random.shuffle(images)

    n_eval = int(len(images) * EVAL_FRACTION)

    selected_images = images[:n_eval]

    # Create corresponding class folder
    eval_class_path = os.path.join(
        eval_path,
        class_name
    )

    os.makedirs(eval_class_path, exist_ok=True)

    for image_name in selected_images:

        source = os.path.join(
            class_path,
            image_name
        )

        destination = os.path.join(
            eval_class_path,
            image_name
        )

        shutil.copy2(source, destination)

    class_counts[class_name] = n_eval

print("Evaluation subset created.")
print("-" * 70)

total = 0

for class_name, count in class_counts.items():
    print(f"{class_name:50s}: {count}")
    total += count

print("-" * 70)
print("TOTAL EVALUATION IMAGES:", total)
print("Location:", eval_path)
```

```python
import os
import shutil
import random

# ============================================================
# TOMATO DATASET V2 — TRAIN / VALIDATION SPLIT
# ============================================================

source_path = "/content/TomatoDatasetV2/extracted/augmented_dataset"

eval_path = "/content/TomatoDatasetV2/evaluation_subset"

train_path = "/content/TomatoDatasetV2/train"
val_path = "/content/TomatoDatasetV2/val"

# Reproducibility
random.seed(42)

# Create folders
os.makedirs(train_path, exist_ok=True)
os.makedirs(val_path, exist_ok=True)

VAL_FRACTION = 0.10

total_train = 0
total_val = 0

print("Creating train/validation split...")
print("-" * 75)

for class_name in sorted(os.listdir(source_path)):

    source_class = os.path.join(
        source_path,
        class_name
    )

    if not os.path.isdir(source_class):
        continue

    # Evaluation images already separated
    eval_class = os.path.join(
        eval_path,
        class_name
    )

    eval_images = set(os.listdir(eval_class))

    # All images
    all_images = [
        f for f in os.listdir(source_class)
        if f.lower().endswith(
            (".jpg", ".jpeg", ".png", ".bmp", ".webp")
        )
    ]

    # Remove evaluation images
    remaining_images = [
        f for f in all_images
        if f not in eval_images
    ]

    # Shuffle remaining images
    random.shuffle(remaining_images)

    # Validation split
    n_val = int(
        len(remaining_images) * VAL_FRACTION
    )

    val_images = remaining_images[:n_val]
    train_images = remaining_images[n_val:]

    # Create class folders
    train_class = os.path.join(
        train_path,
        class_name
    )

    val_class = os.path.join(
        val_path,
        class_name
    )

    os.makedirs(train_class, exist_ok=True)
    os.makedirs(val_class, exist_ok=True)

    # Copy validation images
    for image_name in val_images:

        shutil.copy2(
            os.path.join(source_class, image_name),
            os.path.join(val_class, image_name)
        )

    # Copy training images
    for image_name in train_images:

        shutil.copy2(
            os.path.join(source_class, image_name),
            os.path.join(train_class, image_name)
        )

    total_train += len(train_images)
    total_val += len(val_images)

    print(
        f"{class_name:50s} "
        f"Train={len(train_images):4d} "
        f"Val={len(val_images):3d} "
        f"Eval={len(eval_images):3d}"
    )

print("-" * 75)
print("TOTAL TRAIN:", total_train)
print("TOTAL VAL  :", total_val)
print("TOTAL EVAL :", 3057)

print("\nTrain path:", train_path)
print("Val path  :", val_path)
print("Eval path :", eval_path)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# ============================================================
# LOAD ORIGINAL V4 MODEL
# ============================================================

v4_model = tf.keras.models.load_model(
    "/content/drive/MyDrive/Tomato_Disease_Project/tomato_disease_model_v4.keras"
)

# Get the nested EfficientNet model
base_model = v4_model.get_layer("functional_2")

print("V4 loaded.")
print("Outer model layers:", len(v4_model.layers))
print("Nested model layers:", len(base_model.layers))
print("Input :", v4_model.input_shape)
print("Output:", v4_model.output_shape)

# ============================================================
# FREEZE THE ENTIRE NESTED MODEL FIRST
# ============================================================

base_model.trainable = False

# ============================================================
# UNFREEZE LAST 30 LAYERS
# KEEP BATCH NORMALIZATION FROZEN
# ============================================================

for layer in base_model.layers[-30:]:

    if not isinstance(layer, layers.BatchNormalization):
        layer.trainable = True

# ============================================================
# SHOW TRAINABLE LAYERS
# ============================================================

print("\nTrainable layers inside functional_2:")
print("-" * 75)

trainable_layers = []

for i, layer in enumerate(base_model.layers):

    if layer.trainable:
        trainable_layers.append(layer)

        print(
            f"{i:3d} | "
            f"{layer.name:40s} | "
            f"{layer.__class__.__name__}"
        )

print("-" * 75)

print(
    "\nTrainable nested layers:",
    len(trainable_layers)
)

# ============================================================
# COMPILE
# ============================================================

v4_model.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=1e-5
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# ============================================================
# COUNT TRAINABLE PARAMETERS
# ============================================================

trainable_params = sum(
    tf.size(v).numpy()
    for v in v4_model.trainable_variables
)

total_params = sum(
    tf.size(v).numpy()
    for v in v4_model.variables
)

print("\nModel compiled successfully.")

print(
    "Trainable parameters:",
    f"{trainable_params:,}"
)

print(
    "Total parameters:",
    f"{total_params:,}"
)
```

```python
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau

# ============================================================
# CALLBACKS
# ============================================================

checkpoint_path = "/content/drive/MyDrive/Tomato_Disease_Project/tomato_v2_finetuned_best.keras"

checkpoint = ModelCheckpoint(
    checkpoint_path,
    monitor="val_accuracy",
    save_best_only=True,
    mode="max",
    verbose=1
)

early_stop = EarlyStopping(
    monitor="val_accuracy",
    patience=2,
    mode="max",
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=1,
    min_lr=1e-7,
    verbose=1
)

# ============================================================
# TRAIN
# ============================================================

history_v2 = v4_model.fit(
    v2_train_ds,
    validation_data=v2_val_ds,
    epochs=5,
    callbacks=[
        checkpoint,
        early_stop,
        reduce_lr
    ]
)

print("\nFine-tuning completed.")
print("Best model saved at:")
print(checkpoint_path)
```

```python
import tensorflow as tf
import numpy as np
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

# ============================================================
# LOAD BEST FINE-TUNED MODEL
# ============================================================

best_v2_model = tf.keras.models.load_model(
    "/content/drive/MyDrive/Tomato_Disease_Project/tomato_v2_finetuned_best.keras"
)

print("Best fine-tuned model loaded.")
print("Input :", best_v2_model.input_shape)
print("Output:", best_v2_model.output_shape)

# ============================================================
# PREDICT ON UNTOUCHED EVALUATION SET
# ============================================================

y_true = np.concatenate([
    y.numpy() for _, y in eval_ds
])

y_prob = best_v2_model.predict(
    eval_ds,
    verbose=1
)

y_pred = np.argmax(y_prob, axis=1)

# ============================================================
# BASIC METRICS
# ============================================================

accuracy = accuracy_score(y_true, y_pred)

precision = precision_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

weighted_f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

# ============================================================
# AUC
# ============================================================

auc_macro = roc_auc_score(
    y_true,
    y_prob,
    multi_class="ovr",
    average="macro"
)

auc_weighted = roc_auc_score(
    y_true,
    y_prob,
    multi_class="ovr",
    average="weighted"
)

# ============================================================
# RESULTS
# ============================================================

print("\n" + "=" * 60)
print("TOMATO DATASET V2 — FINE-TUNED MODEL")
print("UNTOUCHED EVALUATION SET")
print("=" * 60)

print(f"Accuracy       : {accuracy * 100:.2f}%")
print(f"Macro Precision: {precision * 100:.2f}%")
print(f"Macro Recall   : {recall * 100:.2f}%")
print(f"Macro F1       : {f1 * 100:.2f}%")
print(f"Weighted F1    : {weighted_f1 * 100:.2f}%")
print(f"Macro AUC      : {auc_macro:.4f}")
print(f"Weighted AUC   : {auc_weighted:.4f}")

print("\nClassification Report:\n")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=eval_ds.class_names,
        digits=4,
        zero_division=0
    )
)

# ============================================================
# CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(y_true, y_pred)

print("\nConfusion Matrix:")
print(cm)
```
